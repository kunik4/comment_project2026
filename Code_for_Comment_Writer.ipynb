```python
import csv

#Make the file with students in the course. The header shows what goes in each collum of the file, which teacher can imput.
file_name = "teacher.csv"

headers = [
    "Name", "Course", "Attributes", "Key Grades", "Work On Learning Outcomes",
    "Suggestions", "Final Grade"
]

students = [{
    "Name": "Tulip",
    "Course": "Chinese III",
    "Attributes": "hard working, resilant",
    "Key Grades": "A's on all tests: 97% on final exam",
    "Work On Learning Outcomes": "tones",
    "Suggestions": "Come to peer language mentors during office hours",
    "Final Grade": "A"
}, {
    "Name": "Daisy",
    "Course": "Chinese III",
    "Attributes":
    "always engaged, curious ",
    "Key Grades": "B's on most tests: 88% on the final exam",
    "Work On Learning Outcomes": "writing and listening",
    "Suggestions":
    "Come by during office hours to study for dictation quizzes",
    "Final Grade": "B+"
}]

rows = []
for student in students:
    row = [student[h] for h in headers]
    rows.append(row)
with open(file_name, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(headers)
    writer.writerows(rows)
```

```python
#name, course, attributes, key grades, learning outcomes, suggestions, sem grade
#file name = student_name.txt
#pep8
#intro is the same for the course

# asks the teacher if they wish to add another student to their file. Then they short response a few questions so that the program can generate a larger comment for them.


def newstudent():
    while True:
        yes_no = input("Would you like to add in another student? (yes/no)"
                       ).strip().lower()
        if yes_no == "yes":
            print("Great! Please answer the questions below.")
            name = input("First Name: ")
            course = input("Course Title: ")
            attributes = input(
                "Attributes (work ethic, behavior, skills, etc.) Please insert two adjectives: "
            )
            grades = input(
                "Over all performance: grade on the final exam. (Please seperate using the colon): "
            )
            outcomes = input(
                "Focus on Learning Outcomes (please only enter in: writing, listening, reading, or speaking): "
            )
            suggestions = input("Suggestions for improvement (the student should...): ")
            final_grade = input("Final Grade in the class: ")

            with open("teacher.csv", "a", newline="") as f:
                writer = csv.writer(f)
                writer.writerow([
                    name, course, attributes, grades, outcomes, suggestions,
                    final_grade
                ])
                print(f"{name} has been added successfully!\n")
        elif yes_no == "no":
            print("Finished adding students.\n")
            break
        else:
            print("Please enter 'yes' or 'no'\n")


#once the teacher has added the amount of kids they desire, then we begin writing the comments for each student.


#this function reads the updated file and returns the data in each collum for each student so we can form the full comment.
def read_csv(file_name):
    """
    Reads the teacher CSV and returns a list of student data
    """
    students = []
    with open(file_name, "r", newline="") as f:
        reader = csv.reader(f)
        headers = next(reader)
        for row in reader:
            students.append(row)
    return students
```

```python
def attributes(stdnt):
    """
    Takes in the attributes of the student,
    returning a sentence with the information incorperated. 
    """
    parts_new = stdnt[2].split(", ")
    adj1 = parts_new[0]
    adj2 = parts_new[1]
    return f"{stdnt[0]}, you are {adj1.strip()} and {adj2.strip()}, making you a joy to have in class. I hope to keep hearing your voice."

    for notes in all_adjs:
        print(notes)


def key_grades(stdnt):
    """
    Takes in the overall performance and the final exam grade of the student,
    returning a sentence with the information incorperated. 
    """
    sections_new = stdnt[3].split(": ")
    overall = sections_new[0]
    final = sections_new[1]
    grade_num = final.split("%")
    is_it_a_ninty = grade_num[0]
    if is_it_a_ninty[0] == "9" or is_it_a_ninty[0:2] == "10":
        return f"Throughout the semester, you recieved {overall}, and finished the year with a {final}."

    else:
        return f"Throughout the semester, you recieved {overall}, and finished the year with an {final}."


def outcome_recs(stdnt):
    """
    Takes in the areas of improvement for the student,
    returning a sentence with the information incorperated. 
    """
    outcome = stdnt[4]
    suggest = stdnt[5]
    if 'and' in outcome:
        return (
            f"For your areas of growth, I suggest that we work on your {outcome}. To improve in these section, you should {suggest.lower()}."
        )
    else:
        return (
            f"For your area of growth, I suggest that we work on your {outcome}. To improve in this section, you should{suggest.lower()}."
        )


def final_grade(stdnt):
    """
    Takes in the final letter grade of the student,
    returning a sentence with the information incorperated. 
    """

    gradde = stdnt[6]

    if "A" in gradde:
        return f"Overall, you will recieve an {gradde} for semester 1. Strong work this year. Keep it up!"
    elif "B" in gradde:
        return f"Overall, you will recieve a {gradde} for semester 1. Strong work this year, let's keep pushing forward and please come to me with any questions."
    elif "F" in gradde:
        return f"Overall, you will recieve an {gradde} for semester 1. Let's keep talking and see what we can do to get you a credit for this course."
    else:
        return f"Overall, you will recieve a {gradde} for semester 1. Good job so far, I really enoy your presence in this class."


def put_it_together(file_name="teacher.csv"):
    stdnt = read_csv(file_name)
    for studs in stdnt:

        intro = f"This semester in {studs[1]}, we have continued building on the foundations of speaking, listening, reading, and writing in Mandarin. Students practiced more complex sentence strutures, expanded their vocabulary, and explored culteral topics through conversations, tests, and projects."

        full_com = " ".join([
            intro,
            attributes(studs),
            key_grades(studs),
            outcome_recs(studs),
            final_grade(studs)
        ])
        print("\n" + full_com + "\n")
        fffname = f"{studs[0]}_comment.txt"
        with open(fffname, "w") as f:
            f.write(full_com)

        print("A file for this indivial student has additionally been made!")
        print()


newstudent()
put_it_together()
```
